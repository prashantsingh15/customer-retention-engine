```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("online_retail_II.csv")
```

```python
df["InvoiceDate"].head(5)
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], errors="coerce")
```

```python
df["InvoiceDate"].dtype
```

```python
df["InvoiceDate"].isna().sum()
```

```python
df = df[df["InvoiceDate"].notna()]

```

```python
df.info()
```

```python
df = df[~df["Invoice"].astype(str).str.startswith("C")]
```

```python
df["Invoice"].head()
```

```python
df = df[(df["Quantity"] > 0) & (df["Price"] > 0)]
```

```python
df[["Quantity", "Price"]].describe()
```

```python
df["total_amount"] = df["Quantity"] * df["Price"]
```

```python
df[["Quantity", "Price", "total_amount"]].head()
```

```python
clean_df = df[
    [
        "Customer ID",
        "Invoice",
        "InvoiceDate",
        "Quantity",
        "Price",
        "total_amount"
    ]
]
```

```python
clean_df.head()
```

```python
clean_df.to_csv(
    "clean_transactions.csv",
    index=False
)
```
