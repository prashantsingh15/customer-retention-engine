```python
import pandas as pd
import sqlite3

# load final churn decision data
sql_df = pd.read_csv("churn_retention_decisions.csv")

# create SQLite database
conn = sqlite3.connect("customer_analytics.db")

# create SQL table
sql_df.to_sql(
    name="churn_retention",
    con=conn,
    if_exists="replace",
    index=False
)

conn.close()
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("customer_analytics.db")

query = """
SELECT 
    SUM(monetary) AS revenue_at_risk
FROM churn_retention
WHERE high_risk = 1;
"""

revenue_risk_df = pd.read_sql(query, conn)
conn.close()

revenue_risk_df
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("customer_analytics.db")

query = """
SELECT 
    retention_action,
    COUNT(*) AS customer_count
FROM churn_retention
GROUP BY retention_action
ORDER BY customer_count DESC;
"""

action_dist_df = pd.read_sql(query, conn)
conn.close()

action_dist_df
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("customer_analytics.db")

query = """
SELECT
    recency,
    frequency,
    monetary,
    churn_probability,
    priority_score,
    retention_action
FROM churn_retention
WHERE high_risk = 1
ORDER BY priority_score DESC
LIMIT 20;
"""

top_customers_df = pd.read_sql(query, conn)
conn.close()

top_customers_df
```
